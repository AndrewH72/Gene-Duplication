```python
# Read in base files.
gtexData = pd.read_parquet("/Users/andrewhsu/Projects/McNair/data/gtexData.parquet")
emtabData = pd.read_parquet("/Users/andrewhsu/Projects/McNair/data/EMTAB6081Data.parquet")
```

```python
filePath = "/Users/andrewhsu/Projects/McNair/notes/tissueSample.tsv"
with open(filePath, "w") as f:
    tissues = gtexData["Tissue.Type"].unique()
    for tissue in tissues:
        samples = gtexData[gtexData["Tissue.Type"] == tissue].index
        f.write(f"{tissue}:\n")
        for sample in samples:
            f.write(f"{sample}\n")
```

```python
# Group the data by the tissue type and then average genes from the same tissue type.
gtexDataReformatted = gtexData.groupby("Tissue.Type").mean().T
gtexDataReformatted.columns.name = None
gtexDataReformatted.index.name = "Gene.ID"
gtexDataReformatted.index = [gene[:gene.find(".")] for gene in gtexDataReformatted.index]
gtexDataReformatted
```

```python
gtexDataReformatted.index.unique().to_series().to_csv("/Users/andrewhsu/Projects/McNair/notes/gtexDataGenes.csv", index=False)
```

```python
# Reformat the name of of columns from XXX_X_Tissue to just Tissue. Then group and average columns with the same name together.
emtabData.columns = [col[re.search(r"[A-Z][a-z]", col).start():] for col in emtabData.columns]
emtabDataReformatted = emtabData.T.groupby(level=0).mean().T
emtabDataReformatted.index.name = "Gene.ID"
emtabDataReformatted = emtabDataReformatted.loc[:, list(gtexDataReformatted.columns)]
emtabDataReformatted
```

```python
emtabDataReformatted.index.unique().to_series().to_csv("/Users/andrewhsu/Projects/McNair/notes/emtabDataGenes.csv", index=False)
```
