```python
# Libraries.
import numpy as np
import pandas as pd
from scipy import stats
```

```python
orthoTable = pd.read_parquet("/Users/andrewhsu/Projects/McNair/data/orthologTableDistDupeBioTypeParentalDaughterGOEPExondNdSProtID.parquet")
randomGenesDF = pd.read_csv("/Users/andrewhsu/Projects/McNair/data/randomGenesTable.csv")
humanParentDaughter = pd.read_parquet("/Users/andrewhsu/Projects/McNair/data/humanParentDaughter.parquet")
mouseParentDaughter = pd.read_parquet("/Users/andrewhsu/Projects/McNair/data/mouseParentDaughter.parquet")

orthoTable = orthoTable[(orthoTable["Human Gene Type"] == "protein_coding") & (orthoTable["Mouse Gene Type"] == "protein_coding")]
orthoTableNoRetro = orthoTable[((orthoTable["Number of Human Exons"] != 1) | (orthoTable["Number of Mouse Exons"] != 1))]
orthoTableNoOverUnderRepresentation = orthoTable[orthoTable["+/-"] == "nan"]
display(orthoTable)

humanParentDaughter = humanParentDaughter.merge(orthoTable[["Gene stable ID", "Mouse gene stable ID", "+/-", "Number of Human Exons", "dN", "dS", "dN/dS"]], left_on=["Mouse Gene", "Human Parent"], right_on=["Mouse gene stable ID", "Gene stable ID"], how="left").drop(columns=["Gene stable ID", "Mouse gene stable ID"]).rename(columns={"Number of Human Exons": "Number of Parental Exons", "+/-": "Parent +/-", "dN": "dN Parent", "dS": "dS Parent", "dN/dS": "dN/dS Parent"})
humanParentDaughter = humanParentDaughter.merge(orthoTable[["Gene stable ID", "Mouse gene stable ID", "+/-", "Number of Human Exons", "dN", "dS", "dN/dS"]], left_on=["Mouse Gene", "Human Daughter"], right_on=["Mouse gene stable ID", "Gene stable ID"], how="left").drop(columns=["Gene stable ID", "Mouse gene stable ID"]).rename(columns={"Number of Human Exons": "Number of Daughter Exons", "+/-": "Daughter +/-", "dN": "dN Daughter", "dS": "dS Daughter", "dN/dS": "dN/dS Daughter"})
mouseParentDaughter = mouseParentDaughter.merge(orthoTable[["Gene stable ID", "Mouse gene stable ID", "+/-", "Number of Mouse Exons", "dN", "dS", "dN/dS"]], left_on=["Human Gene", "Mouse Parent"], right_on=["Gene stable ID", "Mouse gene stable ID"], how="left").drop(columns=["Gene stable ID", "Mouse gene stable ID"]).rename(columns={"Number of Mouse Exons": "Number of Parental Exons", "+/-": "Parent +/-", "dN": "dN Parent", "dS": "dS Parent", "dN/dS": "dN/dS Parent"})
mouseParentDaughter = mouseParentDaughter.merge(orthoTable[["Gene stable ID", "Mouse gene stable ID", "+/-", "Number of Mouse Exons", "dN", "dS", "dN/dS"]], left_on=["Human Gene", "Mouse Daughter"], right_on=["Gene stable ID", "Mouse gene stable ID"], how="left").drop(columns=["Gene stable ID", "Mouse gene stable ID"]).rename(columns={"Number of Mouse Exons": "Number of Daughter Exons", "+/-": "Daughter +/-", "dN": "dN Daughter", "dS": "dS Daughter", "dN/dS": "dN/dS Daughter"})


humanParentDaughterNoRetro = humanParentDaughter[(humanParentDaughter["Number of Parental Exons"] != 1) | (humanParentDaughter["Number of Daughter Exons"] != 1)]
humanParentDaughterNoOverUnderRepresentation = humanParentDaughter[(humanParentDaughter["Parent +/-"] != "nan") | (humanParentDaughter["Daughter +/-"] == "nan")]
mouseParentDaughterNoRetro = mouseParentDaughter[(mouseParentDaughter["Number of Parental Exons"] != 1) | (mouseParentDaughter["Number of Daughter Exons"] != 1)]
mouseParentDaughterNoOverUnderRepresentation = mouseParentDaughter[(mouseParentDaughter["Parent +/-"] != "nan") | (mouseParentDaughter["Daughter +/-"] == "nan")]

parentDaughterDFs = {"Human": [humanParentDaughter, humanParentDaughterNoRetro, humanParentDaughterNoOverUnderRepresentation], "Mouse": [mouseParentDaughter, mouseParentDaughterNoRetro, mouseParentDaughterNoOverUnderRepresentation]}
display(humanParentDaughter)

randomGenesDF = randomGenesDF.merge(orthoTable[["Gene stable ID", "+/-", "Number of Human Exons", "dN", "dS", "dN/dS"]], left_on="HumanID", right_on="Gene stable ID", how="left").drop(columns=["Gene stable ID"]).rename(columns={"+/-": "Human +/-", "dN": "Human dN", "dS": "Human dS", "dN/dS": "Human dN/dS"})
randomGenesDF = randomGenesDF.merge(orthoTable[["Mouse gene stable ID", "+/-", "Number of Mouse Exons", "dN", "dS", "dN/dS"]], left_on="MouseID", right_on="Mouse gene stable ID", how="left").drop(columns=["Mouse gene stable ID"]).rename(columns={"+/-": "Mouse +/-", "dN": "Mouse dN", "dS": "Mouse dS", "dN/dS": "Mouse dN/dS"})

randomGenesNoRetro = randomGenesDF[((randomGenesDF["Number of Human Exons"] != 1) | (randomGenesDF["Number of Mouse Exons"] != 1))]
randomGenesNoOverUnderRepresentation = randomGenesDF[(randomGenesDF["Human +/-"] == "nan") & (randomGenesDF["Mouse +/-"] == "nan")]
display(randomGenesDF)
```

```python
headerCols = ["n", "Median", "Mean", "Rho", "Corr PValue"]
statsDF = pd.DataFrame(
    index = ["one-to-one", "one-to-many",
                "one-to-many, duplicated in human", 
                "one-to-many, duplicated in human, GOC=0", "one-to-many, duplicated in human, GOC=25", "one-to-many, duplicated in human, GOC=50", "one-to-many, duplicated in human, GOC=75", "one-to-many, duplicated in human, GOC=100",
                "one-to-many, duplicated in mouse",
                "one-to-many, duplicated in mouse, GOC=0", "one-to-many, duplicated in mouse, GOC=25", "one-to-many, duplicated in mouse, GOC=50", "one-to-many, duplicated in mouse, GOC=75", "one-to-many, duplicated in mouse, GOC=100",
                "many-to-many", "random",
                "trios, duplicated in human, parental", "trios, duplicated in human, daughter",
                "trios, duplicated in mouse, parental", "trios, duplicated in mouse, daughter"],
    columns=headerCols
)
```

```python
sigDF = pd.DataFrame({
    "Group 1": ["one-to-one", "one-to-one",
                "one-to-many, duplicated in human",
                "many-to-many", "many-to-many",
                "one-to-many, duplicated in human, GOC=0", "one-to-many, duplicated in human, GOC=25", "one-to-many, duplicated in human, GOC=50", "one-to-many, duplicated in human, GOC=75", "one-to-many, duplicated in human, GOC=0",
                "one-to-many, duplicated in mouse, GOC=0", "one-to-many, duplicated in mouse, GOC=25", "one-to-many, duplicated in mouse, GOC=50", "one-to-many, duplicated in mouse, GOC=75", "one-to-many, duplicated in mouse, GOC=0",
                "trios, duplicated in human, parental", "trios, duplicated in mouse, parental"],

    "Group 2": ["one-to-many, duplicated in human", "one-to-many, duplicated in mouse",
                "one-to-many, duplicated in mouse", 
                "one-to-many, duplicated in human", "one-to-many, duplicated in mouse", 
                "one-to-many, duplicated in human, GOC=25", "one-to-many, duplicated in human, GOC=50", "one-to-many, duplicated in human, GOC=75", "one-to-many, duplicated in human, GOC=100", "one-to-many, duplicated in human, GOC=100",
                "one-to-many, duplicated in mouse, GOC=25", "one-to-many, duplicated in mouse, GOC=50", "one-to-many, duplicated in mouse, GOC=75", "one-to-many, duplicated in mouse, GOC=100", "one-to-many, duplicated in mouse, GOC=100",
                "trios, duplicated in human, daughter", "trios, duplicated in mouse, daughter"],
    "P-value": 0.0
})
```

```python
translationTable = {
    "one-to-one": "ortholog_one2one",
    "one-to-many": "ortholog_one2many",
    "duplicated in human": "Human",
    "duplicated in mouse": "Mouse",
    "GOC=0": 0,
    "GOC=25": 25,
    "GOC=50": 50,
    "GOC=75": 75,
    "GOC=100": 100,
    "many-to-many": "ortholog_many2many",
    "random": "random",
    "trios": "ortholog_one2many",
    "duplicated in human": "Human",
    "duplicated in mouse": "Mouse",
    "parental": "Parent",
    "daughter": "Daughter"
}
```

```python
randomGenesList = [randomGenesDF, randomGenesNoRetro, randomGenesNoOverUnderRepresentation]
dfValuesList = []
for idx, df in enumerate([orthoTable, orthoTableNoRetro, orthoTableNoOverUnderRepresentation]):
    dfValues = statsDF.to_dict(orient="index")
    for metric in df.columns[np.r_[4:9, 39:42]]:
        for category in statsDF.index:
            translatedCategory = list(map(translationTable.get, category.split(", ")))
            if "random" in translatedCategory:
                if metric not in df.columns[4:9]:
                    dfValues[category]["n"] = np.nan
                    dfValues[category]["Median"] = np.nan
                    dfValues[category]["Mean"] = np.nan
                    continue
                filteredDF = randomGenesList[idx]
            elif len(translatedCategory) == 1:
                filteredDF = df[df["Mouse homology type"] == translatedCategory[0]]
            elif len(translatedCategory) == 2:
                filteredDF = df[(df["Mouse homology type"] == translatedCategory[0]) & (df["Duplicated Species"] == translatedCategory[1])]
            elif len(translatedCategory) == 3 and isinstance(translatedCategory[2], int):
                filteredDF = df[(df["Mouse homology type"] == translatedCategory[0]) & (df["Duplicated Species"] == translatedCategory[1]) & (df["Mouse Gene-order conservation score"] == translatedCategory[2])]
            elif len(translatedCategory) == 3 and isinstance(translatedCategory[2], str):
                parentalDaughterDF = parentDaughterDFs[translatedCategory[1]][idx]
                filteredDF = parentalDaughterDF.filter(like=translatedCategory[2])
                filteredDF.columns = filteredDF.columns.str.replace(f" {translatedCategory[2]}", "")
            for column in statsDF.columns:
                if column == "n":
                    dfValues[category][column] = filteredDF[metric].dropna().shape[0]
                elif column == "Median":
                    dfValues[category][column] = filteredDF[metric].dropna().median()
                elif column == "Mean":
                    dfValues[category][column] = filteredDF[metric].dropna().mean()
                elif column == "Rho" and len(translatedCategory) > 1 and "one-to-many" in category:
                    filteredTissueNADF = filteredDF[(filteredDF[metric].notna()) & (filteredDF[f"Num {translatedCategory[1]} Dupes"].notna())]
                    dfValues[category][column] = stats.spearmanr(filteredTissueNADF[metric], filteredTissueNADF[f"Num {translatedCategory[1]} Dupes"]).statistic
                elif column == "Corr PValue" and len(translatedCategory) > 1 and "one-to-many" in category:
                    filteredTissueNADF = filteredDF[(filteredDF[metric].notna()) & (filteredDF[f"Num {translatedCategory[1]} Dupes"].notna())]
                    dfValues[category][column] = stats.spearmanr(filteredTissueNADF[metric], filteredTissueNADF[f"Num {translatedCategory[1]} Dupes"]).pvalue
        dfValuesList.append(pd.DataFrame(dfValues).T)
```

```python
sigDFArr= []
for idx, df in enumerate([orthoTable, orthoTableNoRetro, orthoTableNoOverUnderRepresentation]):
    for metric in df.columns[np.r_[4:9, 39:42]]:
        pValueArr = []
        for rowNum in range(sigDF.shape[0]):
            translatedGroup1 = list(map(translationTable.get, sigDF.iloc[rowNum, :].iloc[0].split(", ")))
            translatedGroup2 = list(map(translationTable.get, sigDF.iloc[rowNum, :].iloc[1].split(", ")))
            if len(translatedGroup1) == 1:
                filteredDF1 = df[df["Mouse homology type"] == translatedGroup1[0]]
            elif len(translatedGroup1) == 2:
                filteredDF1 = df[(df["Mouse homology type"] == translatedGroup1[0]) & (df["Duplicated Species"] == translatedGroup1[1])]
            elif len(translatedGroup1) == 3 and isinstance(translatedGroup1[2], int):
                filteredDF1 = df[(df["Mouse homology type"] == translatedGroup1[0]) & (df["Duplicated Species"] == translatedGroup1[1]) & (df["Mouse Gene-order conservation score"] == translatedGroup1[2])]
            elif len(translatedGroup1) == 3 and isinstance(translatedGroup1[2], str):
                parentalDaughterDF1 = parentDaughterDFs[translatedGroup1[1]][idx]
                filteredDF1 = parentalDaughterDF1.filter(like=translatedGroup1[2])
                filteredDF1.columns = filteredDF1.columns.str.replace(f" {translatedGroup1[2]}", "")

            if len(translatedGroup2) == 1:
                filteredDF2 = df[df["Mouse homology type"] == translatedGroup2[0]]
            elif len(translatedGroup2) == 2:
                filteredDF2 = df[(df["Mouse homology type"] == translatedGroup2[0]) & (df["Duplicated Species"] == translatedGroup2[1])]
            elif len(translatedGroup2) == 3 and isinstance(translatedGroup2[2], int):
                filteredDF2 = df[(df["Mouse homology type"] == translatedGroup2[0]) & (df["Duplicated Species"] == translatedGroup2[1]) & (df["Mouse Gene-order conservation score"] == translatedGroup2[2])]
            elif len(translatedGroup2) == 3 and isinstance(translatedGroup2[2], str):
                parentalDaughterDF2 = parentDaughterDFs[translatedGroup2[1]][idx]
                filteredDF2 = parentalDaughterDF2.filter(like=translatedGroup2[2])
                filteredDF2.columns = filteredDF2.columns.str.replace(f" {translatedGroup2[2]}", "")

            pValueArr.append(stats.mannwhitneyu(filteredDF1[metric].dropna(), filteredDF2[metric].dropna())[1])

        sigDF["P-value"] = pValueArr
        sigDFArr.append(sigDF.copy())
```

```python
with pd.ExcelWriter("/Users/andrewhsu/Projects/McNair/data/analysisTable.xlsx") as w:
    for analysisTypeIdx in range(0, 3):
        for metricIdx, metric in enumerate(orthoTable.columns[np.r_[4:9, 39:42]]):
            if metric == "dN/dS":
                metric = "dNdS"
            if analysisTypeIdx == 0:
                dfValuesList[analysisTypeIdx * 5 + metricIdx].to_excel(w, sheet_name=metric, startrow=1, index=True)
                sigDFArr[analysisTypeIdx * 5 + metricIdx].to_excel(w, sheet_name=metric, startrow=1, startcol=8, index=False)
            else:
                dfValuesList[analysisTypeIdx * 5 + metricIdx].to_excel(w, sheet_name=metric, startrow=(21 * analysisTypeIdx + 3 + (analysisTypeIdx - 1)), index=True, header=False)
                sigDFArr[analysisTypeIdx * 5 + metricIdx].to_excel(w, sheet_name=metric, startrow=(21 * analysisTypeIdx + 3 + (analysisTypeIdx - 1)), startcol=8, index=False, header=False)

                sheet = w.sheets[metric]
                sheet.cell(row=24, column=1, value="Retrogenes Removed")
                sheet.cell(row=46, column=1, value="Under/Overrepresentation Removed")
                sheet.cell(row=1, column=1, value="Standard")
                sheet.cell(row=2, column=1, value="Category")
```
